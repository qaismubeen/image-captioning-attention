```python
import sys
sys.path.append("../src")

import pickle
from encoder import build_encoder
from evaluate import load_trained_model, generate_caption

with open("../checkpoints/vocab.pkl", "rb") as f:
    vocab_data = pickle.load(f)

word_to_idx = vocab_data["word_to_idx"]
idx_to_word = vocab_data["idx_to_word"]
max_len = vocab_data["max_len"]

encoder = build_encoder()
decoder = load_trained_model("../checkpoints", vocab_size=len(word_to_idx))

import os
test_images = os.listdir("../data/Images")[:5]

for img_name in test_images:
    img_path = os.path.join("../data/Images", img_name)
    caption, _ = generate_caption(img_path, encoder, decoder, word_to_idx, idx_to_word, max_len)
    print(f"{img_name}: {caption}")
```

```python
import os
import random

all_images = os.listdir("../data/Images")
random.seed(42)
test_sample = random.sample(all_images, 5)

for img_name in test_sample:
    img_path = os.path.join("../data/Images", img_name)
    caption, _ = generate_caption(img_path, encoder, decoder, word_to_idx, idx_to_word, max_len, use_beam_search=True)
    print(f"{img_name}: {caption}")
```
